```python
import polars as pl
from import_dados import * 
```

```python
df_veiculos = read_veiculos()
```

```python
df_veiculos.filter(pl.col('ano_sinistro') == 2025).group_by('tipo_veiculo').agg(pl.count().alias('qtd_veiculos'))
```

```python
faixa_inicio = ((pl.col('ano_fab') - 2001) // 5 * 5 + 2001)

(df_veiculos
    .with_columns(
        pl.when(pl.col('ano_fab').is_between(1900, 1970))
        .then(pl.lit('1900-1970'))
        .when(pl.col('ano_fab').is_between(1971, 2000))
        .then(pl.lit('1971-2000'))
        .when(pl.col('ano_fab') >= 2001)
        .then(pl.format('{}-{}', faixa_inicio, faixa_inicio + 4))
        .otherwise(pl.lit('Fora do intervalo'))
        .alias('faixa_fabricacao')
    )
    .group_by('faixa_fabricacao')
    .agg(pl.len().alias('qtd_veiculos'))
    .sort('qtd_veiculos', descending=True).filter(pl.col('faixa_fabricacao') != "Fora do intervalo")
)
```
